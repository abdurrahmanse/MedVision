# Download and Extract PneumoniaMNIST Images
This notebook bypasses the complex `.npz` storage and directly extracts **real `.png` image files** into your data folder.

This makes it much simpler to view the data, understand the project, and train models normally.

```python
import os
import numpy as np
from PIL import Image
import medmnist
from medmnist import INFO

SEED = 42
SAMPLES_PER_CLASS = 50
RAW_DIR = '../data/raw'

# Ensure our class folders exist
os.makedirs(f"{RAW_DIR}/0", exist_ok=True) # Normal
os.makedirs(f"{RAW_DIR}/1", exist_ok=True) # Pneumonia
```

```python
# 1. Download the highly compressed dataset
info = INFO['pneumoniamnist']
DataClass = getattr(medmnist, info['python_class'])
dataset = DataClass(split='train', download=True, root='../data/')

images = dataset.imgs
labels = dataset.labels.squeeze()
```

```python
# 2. Extract strictly 100 images as physical PNG files
np.random.seed(SEED)

for class_id in [0, 1]:
    class_indices = np.where(labels == class_id)[0]
    chosen_indices = np.random.choice(class_indices, SAMPLES_PER_CLASS, replace=False)
    
    for idx in chosen_indices:
        img_array = images[idx]
        img = Image.fromarray(img_array)
        
        # Save physical images so you can see them!
        img_path = f"{RAW_DIR}/{class_id}/sample_{idx:06d}.png"
        img.save(img_path)

print(f"✅ Successfully extracted {SAMPLES_PER_CLASS * 2} PNG images directly into {RAW_DIR}")
```
