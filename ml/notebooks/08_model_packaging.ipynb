# Phase 14 — Model Packaging

Production inference must know exactly which model and preprocessing configuration it is using. Here we create explicit metadata files to package with our PyTorch `.pt` file so the FastAPI backend knows how to serve it.

```python
import os
import json

MODELS_DIR = '../models/'
os.makedirs(MODELS_DIR, exist_ok=True)

class_mapping = {
    "0": "Normal",
    "1": "Pneumonia"
}

model_config = {
    "model_name": "tiny-cnn",
    "version": "1.0.0",
    "input_size": [1, 28, 28],
    "task": "binary-classification",
    "classes": ["Normal", "Pneumonia"],
    "pytorch_checkpoint": "pneumonia_classifier_v1.pt"
}

preprocessing_config = {
    "format": "grayscale",
    "image_size": [28, 28],
    "pipeline": [
        {"step": "resize", "params": {"height": 28, "width": 28}},
        {"step": "to_tensor"},
        {"step": "normalize", "params": {"mean": [0.5], "std": [0.5]}}
    ]
}

with open(os.path.join(MODELS_DIR, 'class_mapping.json'), 'w') as f:
    json.dump(class_mapping, f, indent=4)

with open(os.path.join(MODELS_DIR, 'model_config.json'), 'w') as f:
    json.dump(model_config, f, indent=4)

with open(os.path.join(MODELS_DIR, 'preprocessing_config.json'), 'w') as f:
    json.dump(preprocessing_config, f, indent=4)

print("✅ Packaging metadata saved successfully in 'ml/models/'")
print("\nGenerated files:")
for filename in ['class_mapping.json', 'model_config.json', 'preprocessing_config.json']:
    print(f"- {filename}")
```
